```python
# When a file is uploaded to S3 → Lambda runs → Lambda triggers a Glue ETL job → Glue processes that file.


import boto3 # type: ignore

glue = boto3.client('glue')      #Hey AWS, I want to use the Glue service from Python.


def lambda_handler(event, context):  #AWS calls this function whenever Lambda is triggered.
    # Get the S3 file path(bucket → which S3 bucket
    #input_key → which file inside the bucket)
    bucket = event['Records'][0]['s3']['bucket']['name']  
    input_key = event['Records'][0]['s3']['object']['key']
    s3_input_path = f's3://{bucket}/{input_key}'  #This is what Glue needs to read the file.(where glue reads this)

    print(f"Triggering Glue job with file: {s3_input_path}")
#Start the Glue job named "automate_etl_support_tickets" and pass the file path as an argument.
    response = glue.start_job_run(                 #
        JobName='automate_etl_support_tickets',  
        Arguments={
            '--input_file_path': s3_input_path
        }
    )
```
